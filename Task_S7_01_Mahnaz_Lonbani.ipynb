## * 1. Receive and convert user information — obtenir_dades()*
1. Ask the user to enter their name, surname and phone number.
2. Ask for years of experience, Catalan level, Excel proficiency and academic qualification.
3. Convert experience to float and the  scores to int.
4. Return all six values if conversion succeeds.
5. If a conversion fails, display an error message and return None.

```python
def obtenir_dades():
  try:
     name_surname= input("Enter your first and last name: ")
     phone_number= input("Enter your phone number: ")
     experience= float(input("Enter your years of experience: "))
     catalan= int(input("Select your Catalan proficiency level: 0 = no knowledge, 1 = up to level, B1 2 = level B2, 3 = C1 or higher "))
     excel= int(input("Select your Excel proficiency level: 0 = basic level, 1 = use of functions, 2 = dynamic tables, 3 = Visual Basic and Macros"))
     academic_title= int(input(" What is your academic title: 0 = ESO, 1 = FP, 2 = University, 3 = Master's degree or higher"))
     return name_surname, phone_number, experience, catalan, excel, academic_title
  except ValueError:
    print ("Invalid numeric input. Please try again.")
    return None
```

### 2. Validate Candidate Information (`validar_dades`)

1. Receive the candidate's name and surname, phone number, years of experience, Catalan level, Excel proficiency, and academic qualification.
2. Check that the name and surname and phone number are not empty.
3. Check that years of experience are greater than or equal to zero.
4. Check that the Catalan, Excel, and academic qualification scores are between 0 and 3, inclusive.
5. Return `True` if all conditions are satisfied.
6. Return `False` if any condition is not satisfied.

```python
def validar_dades(name_surname, phone_number, experience, catalan, excel, academic_title):
  if(name_surname.strip()!= ""
      and phone_number.strip()!= ""
      and experience >=0
      and 0<= catalan<=3
      and 0<= excel<=3
      and 0<= academic_title<=3) :
         return True
  else:
    return False
```

### 3. Calculate the Total Score
1. Receive the candidate's years of experience, Catalan level, Excel proficiency, and academic qualification scores.
2. Multiply the years of experience by 1.5.
3. Add the Catalan, Excel, and academic qualification scores to the weighted experience score.
4. Return the total score.

```python
def calcular_puntuacio(experience, catalan, excel, academic_title ):
  score=(experience *1.5)+ catalan+excel+ academic_title
  return score
```

```python
calcular_puntuacio(2, 2, 2, 2 )
```

### 4. Classify the Candidate

1. Receive the candidate's total score.
2. If the score is greater than or equal to 12, return "High Priority".
3. Otherwise, if the score is greater than or equal to 8, return "Suitable candidate".
4. Otherwise, return "Not a priority".

```python
def classificar_candidat(score):
  if score>= 12:
    return "High Priority"
  elif score>= 8:
    return "Suitable candidate"
  else:
    return "Not a priority"
```

```python
classificar_candidat(9)
```

### 5. Display Individual Candidate Results

1. Receive the candidate's name and surname, total score, and classification.
2. Display the candidate's name and surname.
3. Display the candidate's total score.
4. Display the candidate's classification.

```python
def mostrar_individual(name_surname, score, classification):
  print(f"Candidate: {name_surname}")
  print(f"Total score: {score}")
  print(f"Classification: {classification}")
```

```python
mostrar_individual("mah", 9, "Suitable candidate")
```

### 6. Display the Final Candidate Report

1. Receive the list of registered candidates.
2. Sort the candidates by their total scores in descending order (from highest to lowest).
3. Iterate through the sorted list using a `for` loop.
4. For each candidate, retrieve their name and surname, phone number, and total score from the dictionary.
5. Display these details in the final report.

```python
def mostrar_informe(candidates):
  sorted_candidate= sorted(
      candidates,
      key=lambda candidate:candidate["score"],
      reverse=True
  )
  for candidate in sorted_candidate:
    print(candidate["name_surname"],
          candidate["phone_number"],
          candidate["score"])
```

### 7. Coordinate the Selection Process

1. Create an empty list called `candidates` to store valid candidate information.
2. Start a `while` loop that continues until 10 valid candidates have been registered.
3. Call `obtenir_dades()` to receive and convert the candidate's information.
4. If the function returns `None`, skip the current iteration and request new information.
5. Unpack the six values returned by `obtenir_dades()`.
6. Call `validar_dades()` to check whether the candidate's information meets the required conditions.
7. If the information is invalid, skip the current iteration without adding the candidate to the list.
8. Call `calcular_puntuacio()` to calculate the candidate's total score.
9. Call `classificar_candidat()` to determine the candidate's priority category.
10. Create a dictionary containing the candidate's name, phone number, score, and classification.
11. Append the dictionary to the `candidates` list.
12. Call `mostrar_individual()` to display the candidate's individual result.
13. When 10 valid candidates have been registered, exit the loop.
14. Call `mostrar_informe()` to display the final report sorted by score in descending order.

```python
def main_seleccio():
   candidates=[]
   while len(candidates)< 10:
      data= obtenir_dades()
      if data is None:
        continue
      name_surname, phone_number, experience, catalan, excel, academic_title=data
      is_valid=validar_dades(name_surname, phone_number, experience, catalan, excel, academic_title)
      if not is_valid:
        continue
      score=calcular_puntuacio(experience, catalan, excel, academic_title )
      classification= classificar_candidat(score)
      candidate={"name_surname": name_surname,
                 "phone_number": phone_number,
                 "score": score,
                 "classification": classification}
      candidates.append(candidate)
      mostrar_individual(name_surname, score, classification)
   mostrar_informe(candidates)



```

```python
main_seleccio()
```

 ### 1. Get Behavior Segment and Factor (obtenir_segment_i_factor)

1. Receive the customer's behavior code (codi_comportament) as an input parameter.
2. Create a nested dictionary containing the five behavior codes (0–4).
3. For each behavior code, store the corresponding segment, marketing message, and risk factor.
4. Use the behavior code to retrieve the segment, message, and factor from the dictionary .

```python
def obtenir_segment_i_factor(codi_comportament):
   comportaments={
       0:{
           "segment": "Browsing Only",
           "message": "Explore our products",
           "factor_codi": 0.10
       },
       1: {
           "segment": "Cart Created",
           "message": "Your selected items are waiting for you!",
           "factor_codi": 0.25
       },
       2: {
           "segment": "Checkout Incomplete",
           "message": "You're one step away! Complete your purchase.",
           "factor_codi": 0.50
       },
       3: {
           "segment": "Recently Abandoned Cart",
           "message": "Forgot something? Your cart is still waiting!",
           "factor_codi": 0.75
       },
       4: {
           "segment": "Long-Abandoned Cart",
           "message": "Come back and take another look at your cart!",
           "factor_codi": 0.90
       }
   }
   segment, message, factor_codi= (comportaments[codi_comportament]["segment"],
                              comportaments[codi_comportament]["message"],
                              comportaments[codi_comportament]["factor_codi"])
   return segment, message, factor_codi
```

### 2. Get Age Group Factor (obtenir_factor_edat)

1. Receive the customer's age group (grup_edat) as an input parameter.
2. Create a dictionary containing the four age groups and their corresponding risk factors.
3. Use the customer's age group as a key to retrieve the corresponding factor from the dictionary.
4. Return the age factor.

```python
def obtenir_factor_edat(grup_edat):
  factor_edat={
      "18-25": 1.20,
      "26-40":1.00,
      "41-60":0.85,
      "60+": 0.70
       }
  return factor_edat[grup_edat]
```

### 3. Calculate Final Risk (calcular_risc)

1. Receive the behavior factor (factor_codi) and age factor (factor_edat) as input parameters.
2. Multiply the behavior factor by the age factor.
3. Store the result in a variable named risc_final.
4. Return the final risk score.

```python
def calcular_risc(factor_codi, factor_edat):
  risc_final = factor_codi * factor_edat
  return risc_final
```

### 4. Process Customer Data (processar_clients)

1. Receive the original list of customers (clients).
2. Create an empty list named processed_clients to store the processed results.
3. Use a for loop to iterate through each customer in the list.
4. Extract the customer's ID, name, age, age group, and behavior code.
5. Call obtenir_segment_i_factor() to retrieve the customer's segment, marketing message, and behavior factor.
6. Call obtenir_factor_edat() to retrieve the corresponding age factor.
7. Call calcular_risc() to calculate the final risk score.
8. Create a new dictionary containing the customer's original information and the calculated values.
9. Append the new dictionary to processed_clients.
10. After processing all customers, return the completed list.

```python
def processar_clients(clients):
    processed_clients = []
    for client in clients:
      id_client=client["id_client"]
      name=client["nom"]
      edat=client["edat"]
      grup_edat=client["grup_edat"]
      codi_comportament=client["codi_comportament"]
      segment, message, factor_codi= obtenir_segment_i_factor(codi_comportament)
      factor_edat= obtenir_factor_edat(grup_edat)
      risc_final=calcular_risc(factor_codi, factor_edat)
      processed_client={
                        "id_client": id_client,
                        "name": name,
                        "edat": edat,
                        "grup_edat": grup_edat,
                        "codi_comportament":codi_comportament,
                        "segment": segment,
                        "message":message,
                        "factor_codi": factor_codi,
                        "factor_edat": factor_edat,
                        "risc_final": risc_final
                        }
      processed_clients.append(processed_client)
    return processed_clients



```

###The goal is to transform this data into useful information for doctors. From the raw records, you must:

1. Convert each string into a clean structure (dictionary or similar).

- remove spaces

```python
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

### 1. Split Raw Patient Records (pacients_list)

 Convert raw patient strings into structured lists.

**Algorithm:**

1. Receive the original list of patient records (pacients).
2. Create an empty list named pacients_list.
3. Use a for loop to iterate through each patient record.
4. Split each record using the "|" separator.
5. Append the resulting list to pacients_list.
6. Return the completed list.

```python
def pacients_list(pacients):
  pacients_list=[]
  for x in pacients:
      pacients_list.append(x.split("|"))
  return pacients_list
```

```python
pacients_list(pacients)
```

### 2. Normalize Symptoms (`normalize_symptoms`)

 Clean and standardize patient symptoms.

**Algorithm:**

1. Receive a string containing the patient's symptoms (`symptom`).
2. Create an empty list named `normalized_list`.
3. Split the input string by commas and store the result in `symptom_list`.
4. Use a `for` loop to process each symptom.
5. Remove leading and trailing spaces using `strip()`.
6. Convert the symptom to lowercase using `lower()`.
7. Replace internal spaces with underscores using `replace()`.
8. Append the normalized symptom to `normalized_list`.
9. Return `normalized_list`.

```python
def normalize_symptoms(symptom):
  normalized_list=[]
  symptom_list= symptom.split(",")
  for symptom in symptom_list:
      symptom = symptom.strip().lower().replace(" ", "_")
      normalized_list.append(symptom)
  return normalized_list
```

```python
normalize_symptoms('dolor_cabeza,faTiga')
```

### 3. Calculate Total Risk (`calculate_risk`)
 Calculate the total risk score of a patient based on their symptoms.

**Algorithm:**

1. Receive the normalized symptoms list (`symptoms`) and the risk-level dictionary (`nivells_risc`).
2. Initialize `total_risk` to zero.
3. Use a `for` loop to iterate through each symptom.
4. Retrieve the risk score of the current symptom from `nivells_risc`.
5. Add the score to `total_risk`.
6. After processing all symptoms, return `total_risk`.

```python
def calculate_risk(symptoms, nivells_risc):
  total_risk=0
  for x in symptoms:
    total_risk+=nivells_risc[x]

  return total_risk
```

```python
calculate_risk(['dolor_cabeza', 'fatiga'], nivells_risc)
```

### 4. Create Patient Dictionaries (`pacients_dic`)

 Transform patient records into dictionaries containing cleaned information and total risk scores.

**Algorithm:**

1. Receive the patient records (`pacients`) and the risk-level dictionary (`nivells_risc`).
2. Create an empty list named `pacients_dictionary`.
3. Call `pacients_list()` to split the raw records and store the result in `records`.
4. Use a `for` loop to iterate through each record.
5. Extract the patient's name from the first element.
6. Convert the patient's age to an integer using `int()`.
7. Call `normalize_symptoms()` to normalize the symptoms.
8. Call `calculate_risk()` to calculate the total risk score.
9. Create a dictionary named `paciente` containing the name, age, normalized symptoms, and risk score.
10. Append `paciente` to `pacients_dictionary`.
11. Return the completed list of patient dictionaries.

```python
def pacients_dic(pacients, nivells_risc):
  pacients_dictionary=[]
  records=pacients_list(pacients)
  for record in records:
    paciente={
        "name":record[0],
        "age": int(record[1]),
        "symptom":normalize_symptoms(record[2]),
        "nivel_risk":calculate_risk(normalize_symptoms(record[2]), nivells_risc)
              }
    pacients_dictionary.append(paciente)

  return pacients_dictionary
```

```python
pacients_dic(pacients, nivells_risc)
```

### 5. Sort Patients by Risk (`sort_patients`)

 Prioritize patients according to their calculated risk scores.

**Algorithm:**

1. Receive the list of processed patient dictionaries (`patients_data`).
2. Use `sorted()` to sort the patients based on the `nivel_risk` field.
3. Use a lambda function to access each patient's risk score.
4. Set `reverse=True` to sort the patients from highest to lowest risk.
5. Store the sorted list in `pacients_ordenats`.
6. Return `pacients_ordenats`.

```python
def sort_patients(patients_data):
  pacients_ordenats= sorted(
      patients_data,
      key=lambda record:record['nivel_risk'],
      reverse=True
   )
  return pacients_ordenats
```

```python
patients_data=pacients_dic(pacients, nivells_risc)
sort_patients(patients_data)
```

### 6. Main Function (`main`)
Coordinate patient data processing and risk-based prioritization.

**Algorithm:**

1. Call `pacients_dic()` to process the patient records and calculate their risk scores.
2. Store the processed data in `patients_data`.
3. Call `sort_patients()` to sort the patients by risk score.
4. Store the sorted results in `sorted_patients_data`.
5. Return the sorted patient list.

```python
def main():
  patients_data=pacients_dic(pacients)
  sorted_patients_data= sort_patients(patients_data)
  return sorted_patients_data
```

### 1. Define Character Groups (`get_characters`)

```python
def get_characters():
  import string
  uppercase= string.ascii_uppercase
  lowercase=string.ascii_lowercase
  digits=string.digits
  symbols=string.punctuation
  return uppercase, lowercase, digits, symbols
```

### 2. Select Enabled Character Groups (`select_characters`)

 Select the character groups allowed in the password based on Boolean settings.

**Algorithm:**

1. Receive four Boolean parameters: `majuscules`, `minuscules`, `numeros`, and `signes`.
2. Call `get_characters()` to retrieve the four available character groups.
3. Initialize an empty string named `all_character`.
4. Initialize an empty list named `active_group`.
5. If `majuscules` is `True`, add `uppercase` to both `all_character` and `active_group`.
6. If `minuscules` is `True`, add `lowercase` to both variables.
7. If `numeros` is `True`, add `digits` to both variables.
8. If `signes` is `True`, add `symbols` to both variables.
9. Return `all_character` and `active_group`.

```python
def select_characters(majuscules, minuscules, numeros, signes):
  uppercase, lowercase, digits, symbols=get_characters()
  all_character=""
  active_group=[]
  if majuscules:
    all_character+=uppercase
    active_group.append(uppercase)
  if minuscules:
    all_character+=lowercase
    active_group.append(lowercase)
  if numeros:
    all_character+=digits
    active_group.append(digits)
  if signes:
    all_character+=symbols
    active_group.append(symbols)
  return all_character, active_group

```

### 3. Generate Required Characters (`generate_required`)

 Guarantee at least one character from every enabled group.

**Algorithm:**

1. Receive the list of enabled character groups (`active_group`).
2. Initialize an empty list named `required_characters`.
3. Use a `for` loop to iterate through each active group.
4. Convert each group into a list of individual characters.
5. Use `np.random.choice()` to select one random character from the current group.
6. Append the selected character to `required_characters`.
7. Return `required_characters`.

```python
def generate_required(active_group):
  import numpy as np
  required_characters = []
  for group in active_group:
    required_characters.append(str(np.random.choice(list(group))))
  return required_characters
```

### 4. Complete the Password (`complete_password`)

 Fill the remaining password positions with randomly selected characters.

**Algorithm:**

1. Receive `required_characters`, `all_character`, and `password_length`.
2. Calculate the number of remaining characters by subtracting the length of `required_characters` from `password_length`.
3. Store the result in `remaining_character`.
4. Convert `all_character` into a list.
5. Use `np.random.choice()` to select `remaining_character` additional characters.
6. Add the selected characters to `required_characters` using `extend()`.
7. Return the completed list of characters.

```python
def complete_password(required_characters, all_character, password_length):
    import numpy as np
    remaining_character = password_length - len(required_characters)
    required_characters.extend(
        np.random.choice(
            list(all_character),
            size=remaining_character
        )
    )

    return required_characters
```

### 5. Shuffle and Finalize Password (`shuffle_password`)

 Randomize character positions and convert the password into a string.

**Algorithm:**

1. Receive the completed list of password characters (`required_characters`).
2. Use `np.random.shuffle()` to randomize the order of characters.
3. Convert the shuffled list into a string using `"".join()`.
4. Store the resulting string in `password`.
5. Return `password`.

```python
def shuffle_password(required_characters):
  import numpy as np
  np.random.shuffle(required_characters)
  password="".join(required_characters)
  return password
```

   - `majuscules=True`

2. Call `select_characters()` to obtain `all_character` and `active_group`.
3. Call `generate_required()` to select one random character from each enabled group.

6. Call `shuffle_password()` to randomize the positions and convert the characters into a string.

```python
def create_password(password_length, majuscules=True, minuscules=True,numeros=True,signes=False):
  all_character, active_group= select_characters(majuscules, minuscules, numeros, signes)
  required_characters= generate_required(active_group)
  complete_password(required_characters, all_character, password_length)
  pasword= shuffle_password(required_characters)
  return pasword
```

##2. Algorithm — proccesing_match(lines)
Convert raw match records into structured dictionaries.
1. Create an empty list to store the processed matches.
2. Iterate through each line of the input list.
3. Remove leading and trailing whitespace using strip().
4. Split the line using the tab separator (\t).
5. Extract the home team, away team, and match score.

7. Convert the home and away goals into integers.
8. Create a dictionary containing the home team, away team, home goals, and away goals.

```python
def proccesing_match(lines):
  historic_partits_dic=[]
  for line in lines:
    match= line.strip().split("\t")
    score_team=match[1]
    score=score_team.split("-")
    match_result={
        "home_team": match[0],
        "away_team":match[2],
        "home_goals": int(score [0]),
        "away_goals": int(score[1])
            }
    historic_partits_dic.append(match_result)
  return historic_partits_dic
```

##4. Algorithm — update_team_stats(teams, home_team, away_team, home_goals, away_goals)
 Update the cumulative statistics of both teams after a match.
1. Check whether the home team exists in the teams dictionary.
2. If it does not exist, initialize its statistics.
3. Check whether the away team exists in the teams dictionary.
4. If it does not exist, initialize its statistics.
5. Add the goals scored and conceded to both teams.
6. Compare the home goals and away goals.
7. If the home team wins, increase its wins by one, add three points, and increase the away team's losses by one.
8. If the away team wins, increase its wins by one, add three points, and increase the home team's losses by one.
9. If the match is a draw, increase both teams' draws by one and add one point to each team.
10. Return the updated teams dictionary.

```python
def update_team_stats(teams, home_team, away_team, home_goals, away_goals):
  if home_team not in teams:
     teams[home_team]=initialize_team()

  if away_team not in teams:
     teams[away_team]=initialize_team()

  teams[home_team]["goals scored"] += home_goals
  teams[home_team]["goals conceded"] += away_goals

  teams[away_team]["goals scored"] += away_goals
  teams[away_team]["goals conceded"] += home_goals

  if home_goals>away_goals:
     teams[home_team]["wins"]+=1
     teams[home_team]["total_point"]+=3
     teams[away_team]["losses"]+=1
  elif away_goals>home_goals:
     teams[away_team]["wins"]+=1
     teams[away_team]["total_point"]+=3
     teams[home_team]['losses']+=1
  else:
     teams[home_team]['draws']+=1
     teams[away_team]['draws']+=1
     teams[home_team]["total_point"]+=1
     teams[away_team]["total_point"]+=1
  return teams

```

```python
def calculate_team_stats(matches):
  teams={}
  for match in matches:
    home_team= match["home_team"]
    away_team= match[ "away_team"]
    home_goals= match["home_goals"]
    away_goals= match["away_goals"]
    teams= update_team_stats(teams, home_team, away_team, home_goals, away_goals)
  return teams

```
